```python

# Experiment 3: Transformer-Based Next-Token Prediction
# Using Basic Self-Attention

# 1. Import libraries
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.layers import (
    Input, Embedding, Dense, LayerNormalization,
    MultiHeadAttention, Dropout, GlobalAveragePooling1D
)
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 2. Create sample dataset
texts = [
    "deep learning is powerful",
    "machine learning is useful",
    "deep learning uses neural networks",
    "neural networks learn patterns",
    "transformers use self attention",
    "self attention learns relationships",
    "machine learning uses data",
    "deep neural networks learn",
    "artificial intelligence is growing",
    "deep learning improves prediction",
    "machine learning learns patterns",
    "transformers process natural language",
    "neural networks solve problems",
    "artificial intelligence uses data"
]

# 3. Tokenize text
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)

sequences = tokenizer.texts_to_sequences(texts)
vocab_size = len(tokenizer.word_index) + 1

print("Vocabulary Size:", vocab_size)
print("Vocabulary:", tokenizer.word_index)

# 4. Create input and next-token target pairs
X = []
y = []

for seq in sequences:
    for i in range(1, len(seq)):
        X.append(seq[:i])
        y.append(seq[i])

max_len = max(len(seq) for seq in X)

X = pad_sequences(X, maxlen=max_len, padding="pre")
y = np.array(y)

print("Input Shape:", X.shape)
print("Target Shape:", y.shape)

# 5. Build Transformer model with self-attention
embedding_dim = 32
num_heads = 2

inputs = Input(shape=(max_len,), dtype="int32")

# Word embeddings
embedding = Embedding(
    input_dim=vocab_size,
    output_dim=embedding_dim,
    mask_zero=True
)(inputs)

# Basic self-attention
attention = MultiHeadAttention(
    num_heads=num_heads,
    key_dim=embedding_dim // num_heads
)(
    embedding,
    embedding,
    use_causal_mask=True
)

# Residual connection and normalization
x = LayerNormalization()(embedding + attention)
x = Dropout(0.1)(x)

# Feed-forward layer
ffn = Dense(64, activation="relu")(x)
ffn = Dense(embedding_dim)(ffn)

x = LayerNormalization()(x + ffn)

# Predict next token from the final sequence position
x = tf.keras.layers.Lambda(lambda t: t[:, -1, :])(x)
outputs = Dense(vocab_size, activation="softmax")(x)

model = Model(inputs, outputs)

# 6. Compile model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# 7. Display model architecture
model.summary()

# 8. Train model
history = model.fit(
    X,
    y,
    epochs=100,
    batch_size=16,
    verbose=1
)

# 9. Plot training accuracy
plt.plot(history.history["accuracy"])
plt.title("Transformer Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.grid(True)
plt.show()

# 10. Plot training loss
plt.plot(history.history["loss"])
plt.title("Transformer Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()

# 11. Predict the next token
def predict_next_token(text):
    seq = tokenizer.texts_to_sequences([text])

    if not seq[0]:
        return "Unknown input"

    seq = pad_sequences(
        seq,
        maxlen=max_len,
        padding="pre",
        truncating="pre"
    )

    prediction = model.predict(seq, verbose=0)
    predicted_id = int(np.argmax(prediction[0]))

    for word, index in tokenizer.word_index.items():
        if index == predicted_id:
            return word

    return "Unknown"

# 12. Test predictions
test_inputs = [
    "deep learning",
    "machine learning",
    "neural networks",
    "artificial intelligence"
]

for text in test_inputs:
    print("Input:", text)
    print("Predicted next token:", predict_next_token(text))
    print()

print("Next-token prediction completed!")
```
